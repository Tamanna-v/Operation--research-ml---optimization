# ML-Based Demand Forecasting and Inventory Optimization

**Purpose:** Demonstrate the pipeline `historical demand → ML forecast → OR inventory decision`.

**Important:** The demonstration data below are synthetic. They are included only to make the code reproducible. They are **not** empirical research results. For a PhD-ready version, replace them with a documented public dataset and rerun all experiments.


```python
import sys, os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
sys.path.append('../src')
from forecasting import train_random_forest
from inventory_optimization import optimize_order_quantity
```

```python
rng = np.random.default_rng(42)
dates = pd.date_range('2024-01-01', periods=365, freq='D')
seasonal = 20 + 5*np.sin(2*np.pi*dates.dayofyear/365)
weekday = np.where(dates.dayofweek < 5, 3, -2)
demand = np.maximum(1, np.round(100 + seasonal + weekday + rng.normal(0, 8, len(dates))))
df = pd.DataFrame({'date': dates, 'demand': demand})
df.head()
```

```python
model, metrics, predictions = train_random_forest(df)
metrics
```

```python
plt.figure(figsize=(10,4))
plt.plot(predictions['date'], predictions['demand'], label='Actual')
plt.plot(predictions['date'], predictions['predicted_demand'], label='Predicted')
plt.title('Synthetic demonstration: demand forecast')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
predicted = float(predictions['predicted_demand'].iloc[-1])
decision = optimize_order_quantity(predicted_demand=predicted, current_inventory=20)
decision
```
